```python
#2.2
RATE =44100

def sine(freq, rate=RATE,duration=1,amplitude=1) : #on met des valeurs défauts si on a pas la donnée
    abs = np.linspace(0,1,rate)
    ord = np.sin(2*np.pi*freq*abs)
    plt.plot(abs, ord)
    plt.show()
    display(Audio(ord,rate=RATE) )

sine(440)
```

```python
#2.3
RATE = 44100
def sine_linear(freq1,freq2,duration=1,amplitude=1) :
    varfreq =np.linspace(freq1,freq2,RATE)
    abs = np.linspace(0,1,RATE)
    ord = np.sin(2*np.pi*varfreq*abs)
    plt.plot(abs, ord)
    plt.show()
    display(Audio(ord,rate=RATE) )

sine_linear(200,440)
```
